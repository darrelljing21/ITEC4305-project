### Problem 
Rental markets are characterized by significant price dispersion across units that differ in structural features, amenities, and location. However, property managers and renters often lack a transparent, data-driven benchmark for determining whether a listing is fairly priced relative to comparable units.


### Stakeholder: 
<ul><li>Property managers seeking to optimize rental returns</li>
<li>Landlords aiming to minimize vacancy risk through competitive pricing</li>
<li>Renters attempting to identify high-value opportunities</li></ul>

### Decision: Determine the fair market value of a rental 

listing based on its physical and spatial attributes, and identify whether it is underpriced or overpriced relative to comparable units?


### Why This Matters
Mispriced units create economic inefficiencies:
<ul><ol>- Overpriced units may experience longer vacancy periods</ol>
<ol>-Underpriced units represent forgone revenue</ol>
<ol>-Lack of transparent benchmarks reduces negotiation confidence</ol></ul>
A data-driven fair value framework enables more informed pricing and rental decisions.

### Proposal Objectives
<ul><li>Build a Pricing Model: Create a model that predicts price while standardizing for price_type (e.g., converting all to a Monthly baseline).</li>
<li>Feature Engineering: Extract spatial signals from latitude and longitude and categorical value from pets_allowed and amenities.</li>
<li>Define Actionable Price Bands: Create "Deal," "Fair," and "Premium" labels based on the model's confidence intervals.</li>
<li>Model Selection: Compare candidate models (e.g., Random Forest vs. XGBoost) using RMSE to ensure the most reliable valuation.</li></ul>


### Initial Hypothesis
<ul><li>Location: Precise coordinates (latitude/longitude) will be the strongest predictors, outperforming general cityname or state attributes.</li>
<li>The "Pet Premium": Units where pets_allowed is true will command a statistically significant price premium across all square footage levels.</li>
<li>Non-linear interaction: The relationship between square_feet and price is not linear; larger units often have a diminishing price-per-square-foot, which nonlinear models will capture better.</li></ul>



### Success Metric: 
- **Primary Modeling Metric: MAE (Mean Absolute Error) or RMSE in dollars.**
- **Operational Metric: Useful separation of "Underpriced" vs "Overpriced" listings to identify "Market Deals"**
- **Business Meaning: Reduced "Days on Market" for listings priced within the model's suggested Fair Value range.**

### Dataset introduction and EDA summary 
- **The apartment for rent classified dataset contains 10,000 records with 22 columns. Each row representing one apartment is listed for rental with a unique id assigned. It includes features such as rental price, number of bedrooms, number of bathrooms, square feet, location, listing source, etc.**
- **The dataset contains missing values for amenities, bathrooms, bedrooms, pets allowed, address, cityname, state, latitude and longitude.**
