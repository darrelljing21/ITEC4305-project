- **The dataset contains 10,000 apartment listings with 22 variables. The target variable price has no missing values. Most key numeric predictors (bedrooms, bathrooms, square_feet, latitude, longitude) have very low missing rates (<1%), which can be handled with simple imputation. However, several text-based fields have substantial missingness, including pets_allowed (41.6%), amenities (35.5%), and address (33.3%). To keep the project decision-focused and avoid heavy NLP, we will primarily use structured housing and location variables, while treating high-missing text fields as optional features.**

```python
use_cols = [
    "price", "bedrooms", "bathrooms", "square_feet",
    "cityname", "state", "latitude", "longitude",
    "has_photo", "fee", "category", "pets_allowed"
]
df_model = df[use_cols].copy()
df_model.head()
```

```python
import numpy as np
df_model["bedrooms"] = df_model["bedrooms"].fillna(df_model["bedrooms"].median())
df_model["bathrooms"] = df_model["bathrooms"].fillna(df_model["bathrooms"].median())

df_model["cityname"] = df_model["cityname"].fillna("Unknown")
df_model["state"] = df_model["state"].fillna("Unknown")

df_model["latitude"] = df_model["latitude"].fillna(df_model["latitude"].median())
df_model["longitude"] = df_model["longitude"].fillna(df_model["longitude"].median())

df_model["pets_allowed"] = df_model["pets_allowed"].fillna("No")
df_model["pet_friendly"] = np.where(df_model["pets_allowed"].str.strip().str.lower() == "no", 0, 1)
```

```python
import pandas as pd

unit = "ms" if df["time"].max() > 1e12 else "s"

df["datetime"] = pd.to_datetime(df["time"], unit=unit, utc=True) 
df["datetime_local"] = df["datetime"].dt.tz_convert("America/Toronto") 
```

```python
df["datetime_str"] = df["datetime_local"].dt.strftime("%Y-%m-%d %H:%M:%S")
df[["time", "datetime_local", "datetime_str"]].head()
```

```python
df["year"] = df["datetime_local"].dt.year
df["month"] = df["datetime_local"].dt.month
df["day"] = df["datetime_local"].dt.day
df["hour"] = df["datetime_local"].dt.hour
df["weekday"] = df["datetime_local"].dt.day_name()

df[["datetime_str", "year", "month", "day", "hour", "weekday"]].head()
```

```python
unit = "ms" if df["time"].max() > 1e12 else "s"

df["datetime"] = pd.to_datetime(df["time"], unit=unit)

print("Earliest time:", df["datetime"].min())
print("Latest time:", df["datetime"].max())
```

```python
import matplotlib.pyplot as plt

df["year_month"] = df["datetime"].dt.to_period("M")
monthly_counts = df.groupby("year_month").size()

monthly_counts.index = monthly_counts.index.to_timestamp()

plt.figure()
plt.plot(monthly_counts.index, monthly_counts.values)
plt.xlabel("Time")
plt.ylabel("Number of Listings")
plt.title("Monthly Listings Over Time")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
